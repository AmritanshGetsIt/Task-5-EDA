```python
# Titanic EDA - Import Libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("train.csv")

# Basic Exploration
print(df.info())
print(df.describe())
print(df.isnull().sum())
print(df['Survived'].value_counts())

# Visual: Survival Count
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()

# Visual: Sex vs Survival
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival by Sex")
plt.show()

# Visual: Class vs Survival
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Survival by Passenger Class")
plt.show()

# Visual: Age Distribution
plt.hist(df['Age'].dropna(), bins=30)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

# Visual: Boxplot of Age vs Pclass
sns.boxplot(x='Pclass', y='Age', data=df)
plt.title("Passenger Age by Class")
plt.show()

# Heatmap for Missing Data
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Data Heatmap")
plt.show()

# Pairplot (cleaned subset)
sns.pairplot(df[['Survived', 'Pclass', 'Age', 'Fare']], hue='Survived')
plt.show()

# Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation")
plt.show()
```
